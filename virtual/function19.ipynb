##### 온도를 변환하는 프로그램 : 구조화 프로그램

```python
def menu():
    print('1. 섭씨 온도 -> 화씨 온도')
    print('2. 화씨 온도 -> 썹씨 온도')
    print('3. 종료')
    selection = int(input('메뉴를 선택하세요.'))
    return selection

def ctof(c):
    temp = c * 0.9 / 5.0 + 32
    return temp

def ftoc(f):
    temp = (f - 32.0) * 5.0 / 9.0
    return temp

def input_f():
    f = float(input('화씨 온도를 입력하시오.'))
    return f

def input_c():
    c = float(input("섭씨 온도를 입력하시오."))
    return c

def main():
    while True:
        index = menu()

        if index == 1:
            t = input_c()
            t2 = ctof(t)
            print(f'화씨 온도 : {t2}')
        elif index == 2:
            t = input_f()
            t2 = ftoc(t)
            print(f'섭씨 온도 : {t2}')
        else:
            break


main()
```

```python
# 사용자가 선택하는 도형의 면적을 계산하는 프로그램
# 예) 1. 삼각형 2. 사각형 3. 사다리꼴 4. 원
# 면적을 계산할 도형을 선택하세요.

def selection():
    user_selction = input("면적을 계산할 도형을 선택하세요.")
    return user_selction

def triangle():
    user_width = int(input('가로 길이를 작성해주세요.'))
    user_height = int(input('세로 길이를 작성해주세요.'))
    area = user_width * user_height / 2
    print(f'삼각형의 넓이는 {area}입니다.')

def quadrangle():
    user_width = int(input('가로 길이를 작성해주세요.'))
    user_height = int(input('세로 길이를 작성해주세요.'))
    area = user_width * user_height
    print(f'사각형의 넓이는 {area}입니다.')

def trapezoid():
    user_width = int(input('윗변 길이를 작성해주세요.'))
    user_height = int(input('아랫변 길이를 작성해주세요.'))
    user_depth = int(input('높이를 작성해주세요.'))
    area = (user_width+user_height)*user_depth/2
    print(f'사다리꼴의 넓이는 {area}입니다.')

def round():
    user_radius = int(input('반지름을 작성해주세요.'))
    area = user_radius*3.14
    print(f'원의 넓이는 {area}입니다.')

def main():
    while True:
        user_choice = selection()
        if user_choice == '1':
            triangle()
        elif user_choice == '2':
            quadrangle()
        elif user_choice == '3':
            trapezoid()
        elif user_choice == '4':
            round()
        else:
            break

main()
```
